## Data Cleaning

```python
backup = df.copy() # when working with big datasets, data cleaning may corrupt the dataset and reloading again and again takes a lot of time
```

```python
df.head()
```

```python
df['year'].unique()
```

We will filter out the numeric values and discard the rest

```python
df = df[df['year'].str.isnumeric()]
```

```python
df['year'].unique()
```

```python
# it's still object

df['year'] = df['year'].astype(int)
```

```python
df.info()
```

```python
df.shape
```

```python
backup.shape
```

```python
df['company'].unique()
```

```python
df['price'] # it had "Ask For Price" col, we will exclude it manually
```

```python
df = df[df['price'] != 'Ask For Price']
```

```python
df['price']
```

```python
# now we replace the comma with '', then convert it into int

df['price'] = df['price'].str.replace(",", "").astype(int)
```

```python
df.info()
```

```python
df['kms_driven'] # we'll remove kms, then commas, then keep only numeric values and convert it to string
```

```python
df['kms_driven'] = df['kms_driven'].str.split(" ").str.get(0).str.replace(",", "")
```

```python
df= df[df['kms_driven'].str.isnumeric()]
```

```python
df['kms_driven']
```

```python
df['kms_driven'] = df['kms_driven'].astype(int)

df.info()
```

```python
df = df[~df['fuel_type'].isnull()]
```

```python
df['name']
```

```python
# we will extract first 3 words from the name

df['name'] = df['name'].str.split(" ").str.slice(0,3).str.join(" ")
```

```python
df
```

```python
# Since a lot of rows got deleted, reset index is needed

df = df.reset_index(drop=True)
```

```python
df
```

```python
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.describe()
```

Here, the 3rd quartile is almost 5 lakhs, but our maximum value is 85 Lakhs. That means we have an outlier.

```python
df=df[df['price']<6e6]
```

```python
df.reset_index(drop=True, inplace=True)
```
