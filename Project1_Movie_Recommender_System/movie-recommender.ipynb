'[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'

this is a string of list

There is a module called <code>**ast**</code> in python, which does this exact evaluation.
Similar to <code>eval</code>, but safer.

```python
# helper function
import ast

def convert(obj):
    li = []
    for o in ast.literal_eval(obj):
        li.append(o['name'])
    return li
```

```python
movies['genres'].apply(convert)
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies['keywords'].head(1)
```

```python
def convert_cast(obj):
    li = []
    counter = 0
    
    for o in ast.literal_eval(obj):
        if counter != 3: # we only need top 3 actors
            li.append(o['name'])
            counter += 1
        else:
            break
    return li
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies.head(1)
```

```python
movies['crew'][0]
```

```python
# to extract names of director from crew members
def extract_director(obj):
    d = []
    for o in ast.literal_eval(obj):
        if o['job'] == "Director":
            d.append(o['name']) # we want these in list, so that we can combine all them for our Tags field
            break
    return d
```

```python
movies['crew'].apply(extract_director)
```

```python
# Converting the overview field in a list
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
# Removing spaces from sentences which acts as a single entity in a list

movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", '') for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", '') for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", '') for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", '') for i in x])
```

```python
movies.head()
```

```python
movies["tags"] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies['tags'].values
```

```python
movies_new_df = movies[['id',"title", 'tags']]
```

```python
movies_new_df["tags"] = movies_new_df["tags"].apply(lambda x:" ".join(x))
```
